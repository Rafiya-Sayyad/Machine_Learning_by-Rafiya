```python
# 1. Import Statements
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression

# success
```


Pandas:	Handles our data
Matplotlib: Shows our data as graphs
train_test_split: Divides data for learning and testing
LinearRegression: Creates the prediction model
Scikit-learn: The ML library that provides train_test_split and LinearRegression

```python
# 2. Load the Dataset

Data = pd.read_excel("C:/Users/rafiy/Documents/Test_programs/ML/Example1.xlsx")
print(Data)
```

```python
# Explore the dataset

print("\nFirst 5 rows:")
print(Data.head())

print("\nLast 5 rows:")
print(Data.tail())
```

head(),tail(): are methods

```python

print("\nDataset Information:")
print(Data.info())

print("\nStatistical Summary:")
print(Data.describe())
```

Data.info(): Structure of the dataset
Data.describe(): Statistics of the dataset

```python
# 4. Splitting the Data into X and y

# X = Independent Variable / Feature
x = Data[["Hours_studied(x)"]]
# y = Dependent Variable / Target
y = Data["Exam_Scored(y)"]

```

```python
# Printing x values

print("\nx: Hours_studied(x)")
print(x)
```

```python
print("\ny - Exam_Scored(y):")
print(y)
```

```python
# 5. Create the Linear Regression Model

model = LinearRegression()
```

```python
# 6. Train the Model

model.fit(x, y)
```

```python
# 8. Predict Salary

Studied = [[18]]

final = model.predict(Studied)

print("\nPredicted Studied hours:")
print(final)
```

```python
from sklearn.metrics import r2_score
```

```python
# 9. Evaluate the Model
y_pred = model.predict(x)

r2 = r2_score(y, y_pred)

print("R²  :", r2*100, "%")
```

```python
# 10. Plot the Data and Regression Line

plt.scatter(
    x,
    y,
    label="Actual Data"
)

plt.plot(
    x,
    y_pred,
    label="Regression Line"
)

plt.xlabel("Hours_studied")
plt.ylabel("Exam_Scored")
plt.title("Linear Regression - Studied prediction")

plt.legend()
plt.show()
```
