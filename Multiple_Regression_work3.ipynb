```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
```

```python
Data=pd.read_excel("C:/Users/rafiy/Documents/ML_BY_Rafiya/smart_city_traffic_stress_dataset.csv.xlsx")
print(Data)
```

```python
print("\nLast 5 rows:")
print(Data.tail())
```

```python
print("\nData Information:")
print(Data.info())
```

```python
print("\nStatistical information:")
print(Data.describe())
```

```python
Data["weather_condition(Y)"] = Data["weather_condition(Y)"].map({
    "Foggy": 1,
    "Hot": 2,
    "Clear":3,
    "Rainy":4
})
Data["driver_experience_level(Y)"] = Data["driver_experience_level(Y)"].map({
    "Intermediate": 1,
    "Beginner": 2,
    "Expert": 3
})
```

```python
Y=Data[["traffic_density(Y)","horn_events_per_min(Y)","avg_speed(Y)","signal_wait_time(Y)","weather_condition(Y)","road_quality_score(Y)","driver_experience_level(Y)"]]
```

```python
print("\nY-Input:")
print(Y)
```

```python
Z=Data[["stress_index(Z)"]]
```

```python
print("\nZ-Output:")
print(Z)
```

```python
model=LinearRegression()
```

```python
model.fit(Y,Z)
```

```python
print(Data.max())
```

```python
stress=[[118.00, 23.45, 67.00, 73.00, 3.00, 9.00, 2.00]]
final = model.predict(stress)

print("\nPredicted stress:")
print(final)
```

```python
stress2=[[120.00,25.00,92.00,75.50,5.00,11.00,4.00]]
final = model.predict(stress2)

print("\npredicted stress2:")
print(final)
```

```python
Z_pred = model.predict(Y)

r2 = r2_score(Z, Z_pred)

print("R2 :", r2*100, "%")
```
